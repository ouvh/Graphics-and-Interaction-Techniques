# Image processing: transformation and filtering

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
# Load the image
image = cv2.imread('your_image.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convert to RGB format
gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)  # Convert to grayscale
```

```python

# 2.1 Linear Transformations
# Brightness adjustment
brightness_factor = 50
brightened = cv2.add(gray, brightness_factor)

# Contrast stretching
min_val = gray.min()
max_val = gray.max()
contrast_stretched = (gray - min_val) * (255.0 / (max_val - min_val))
contrast_stretched = contrast_stretched.astype(np.uint8)

# 2.2 Image Filtering
# Mean filtering
mean_filtered = cv2.blur(gray, (5, 5))

# Gaussian filtering
gaussian_filtered = cv2.GaussianBlur(gray, (5, 5), 1.5)

# 2.3 Global Thresholding
_, global_thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

# 2.4 Edge Detection (Sobel)
sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_magnitude = np.sqrt(sobelx**2 + sobely**2)
sobel_magnitude = np.uint8(sobel_magnitude / sobel_magnitude.max() * 255)

# 2.5 Histogram Equalization
equalized = cv2.equalizeHist(gray)

# 2.6 2D Convolution
sharpening_kernel = np.array([[-1, -1, -1],
                              [-1, 9, -1],
                              [-1, -1, -1]])
sharpened = cv2.filter2D(gray, -1, sharpening_kernel)

# Display results
plt.figure(figsize=(15, 15))

plt.subplot(3, 3, 1); plt.imshow(gray, cmap='gray'); plt.title('Original Gray')
plt.subplot(3, 3, 2); plt.imshow(brightened, cmap='gray'); plt.title('Brightness Adjusted')
plt.subplot(3, 3, 3); plt.imshow(contrast_stretched, cmap='gray'); plt.title('Contrast Stretched')
plt.subplot(3, 3, 4); plt.imshow(mean_filtered, cmap='gray'); plt.title('Mean Filtered')
plt.subplot(3, 3, 5); plt.imshow(gaussian_filtered, cmap='gray'); plt.title('Gaussian Filtered')
plt.subplot(3, 3, 6); plt.imshow(global_thresh, cmap='gray'); plt.title('Global Thresholding')
plt.subplot(3, 3, 7); plt.imshow(sobel_magnitude, cmap='gray'); plt.title('Sobel Edges')
plt.subplot(3, 3, 8); plt.imshow(equalized, cmap='gray'); plt.title('Histogram Equalized')
plt.subplot(3, 3, 9); plt.imshow(sharpened, cmap='gray'); plt.title('Sharpened')

plt.tight_layout()
plt.show()

plt.show()
```
